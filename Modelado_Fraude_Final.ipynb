## 2b. Ingeniería de variables: features de velocidad (Inciso 2)

Se crean cuatro features de comportamiento temporal por cliente que capturan patrones
de velocidad anómala — indicadores clave de fraude por uso ilegítimo de tarjeta.

| Feature | Descripción |
|---------|-------------|
| `time_since_last_txn_min` | Minutos desde la última transacción del mismo cliente |
| `txn_count_last_1h` | Transacciones del cliente en la última hora (ventana móvil) |
| `txn_count_last_24h` | Transacciones del cliente en las últimas 24 horas (ventana móvil) |
| `amount_zscore_customer` | Z-score del monto vs histórico del cliente *(calculado en Sección 2c)* |

> **Nota técnica — data leakage:** `amount_zscore_customer` se calcula **después del split temporal**
> para que la media y desviación del cliente se computen solo con transacciones de entrenamiento
> (Ene–May) y no con datos de Junio (test).

```python
import pandas as pd

# Cargar columnas temporales y de cliente desde bolivia_limpio.csv
df_limpio = pd.read_csv(
    'data_limpia/bolivia_limpio.csv',
    usecols=['DE7_transmission_datetime', 'client_id', 'amount_usd']
)

# Parsear timestamp: DE7 es MMDDHHmmss guardado como int sin leading zero
# Rango del dataset: 2025-01-01 a 2025-06-30
df_limpio['txn_dt'] = pd.to_datetime(
    '2025' + df_limpio['DE7_transmission_datetime'].astype(str).str.zfill(10),
    format='%Y%m%d%H%M%S'
)

# ── Feature 1: time_since_last_txn_min ───────────────────────────────────────
# Tiempo en minutos desde la transaccion anterior del mismo cliente (orden cronologico)
time_diff = (
    df_limpio.sort_values(['client_id', 'txn_dt'])
             .groupby('client_id')['txn_dt']
             .diff()
             .dt.total_seconds()
             .div(60)
             .reindex(df_limpio.index)       # restaurar orden original del CSV
)
# Primera transaccion del cliente -> sin referencia previa: imputar con mediana global
median_time = time_diff.median()
df_limpio['time_since_last_txn_min'] = time_diff.fillna(median_time)

# ── Features 2 & 3: txn_count_last_1h / txn_count_last_24h ──────────────────
# Ventana rodante por cliente: cuenta transacciones PREVIAS en la ventana de tiempo.
# closed='left' excluye la transaccion actual (no se auto-cuenta)
results_1h, results_24h = [], []
for client, group in df_limpio.groupby('client_id'):
    g = group.sort_values('txn_dt').set_index('txn_dt')
    r1  = g['amount_usd'].rolling('60min', closed='left').count().fillna(0).astype(int)
    r24 = g['amount_usd'].rolling('24H',   closed='left').count().fillna(0).astype(int)
    sorted_idx = group.sort_values('txn_dt').index
    results_1h.append( pd.Series(r1.values,  index=sorted_idx))
    results_24h.append(pd.Series(r24.values, index=sorted_idx))

df_limpio['txn_count_last_1h']  = pd.concat(results_1h)
df_limpio['txn_count_last_24h'] = pd.concat(results_24h)

NEW_FEATURES = ['time_since_last_txn_min', 'txn_count_last_1h', 'txn_count_last_24h']
print('Features de velocidad calculadas:\n')
print(df_limpio[NEW_FEATURES].describe().round(2))
```

## 2c. Split temporal: Enero–Mayo (train) vs Junio 2025 (test)

El PDF especifica: *"Se usará como test de evaluación el mes de Junio 2025"*.

El split temporal es metodológicamente correcto para series de tiempo financieras:
el modelo se entrena en transacciones pasadas y se evalúa en transacciones futuras,
replicando el escenario real de producción. `amount_zscore_customer` se calcula aquí
usando **solo estadísticas del período de entrenamiento** para evitar data leakage.

```python
# Identificar filas de train (Ene-May) y test (Jun)
is_train = df_limpio['txn_dt'].dt.month.isin([1, 2, 3, 4, 5])
is_test  = df_limpio['txn_dt'].dt.month == 6

idx_train = df_limpio.index[is_train]
idx_test  = df_limpio.index[is_test]

print(f'Train (Ene-May 2025): {is_train.sum():,} filas  ({is_train.mean()*100:.1f}%)')
print(f'Test  (Jun 2025):     {is_test.sum():,} filas  ({is_test.mean()*100:.1f}%)')

# ── amount_zscore_customer: stats del cliente calculados SOLO sobre train ─────────────
# Evita data leakage: el test set no influye en la normalizacion del historico del cliente
client_stats = (
    df_limpio.loc[is_train]
    .groupby('client_id')['amount_usd']
    .agg(client_mean='mean', client_std='std')
)
df_limpio = df_limpio.join(client_stats, on='client_id')
df_limpio['amount_zscore_customer'] = (
    (df_limpio['amount_usd'] - df_limpio['client_mean'])
    / (df_limpio['client_std'].fillna(1) + 1e-10)
)
df_limpio.drop(columns=['client_mean', 'client_std'], inplace=True)

ALL_NEW_FEATURES = NEW_FEATURES + ['amount_zscore_customer']

# Combinar con model_ready: mismo origen y mismo orden de filas -> join por indice directo
df_full = df_model.join(df_limpio[ALL_NEW_FEATURES])

print()
for nombre, idx in [('Train', idx_train), ('Test ', idx_test)]:
    subset = df_full.loc[idx, 'is_fraud_binary']
    print(f'{nombre} — Fraude: {subset.sum():,} ({subset.mean()*100:.2f}%)'
          f'  |  No Fraude: {(subset == 0).sum():,}')
```

## 3b. Mejora 1 — SMOTE: sobremuestreo sintético del conjunto de entrenamiento

**¿Por qué SMOTE y no otro método?**

El dataset presenta un desbalance severo (~19:1). SMOTE (*Synthetic Minority Over-sampling Technique*)
genera muestras sintéticas interpolando entre ejemplos reales de fraude, en lugar de simplemente
duplicarlos (oversampling ingenuo). Esto expone al modelo a una región más amplia del espacio de
features para la clase minoritaria.

**Regla de oro — dónde aplicarlo:**
SMOTE se aplica **únicamente sobre `X_train` / `y_train`**, NUNCA sobre el test set.
Aplicarlo sobre test produciría data leakage: el modelo se evaluaría con datos sintéticos
que no existen en producción, inflando artificialmente las métricas.

> **Nota sobre columnas categóricas:** se usa `SMOTENC` (variante de SMOTE para datos mixtos)
> que maneja correctamente las columnas categóricas presentes en el dataset.


```python
# pip install imbalanced-learn -q
from imblearn.over_sampling import SMOTENC, SMOTE
from sklearn.impute import SimpleImputer
import numpy as np

print('=== SMOTE sobre conjunto de entrenamiento ===')
print(f'Shape original X_train: {X_train.shape}')
print(f'Fraudes en train antes de SMOTE: {y_train.sum():,} ({y_train.mean()*100:.2f}%)')

# ── Paso 1: Codificar columnas categóricas a enteros ──────────────────────────
X_train_sm = X_train.copy()
for col in cat_cols:
    X_train_sm[col] = X_train_sm[col].cat.codes

# ── Paso 2: Imputar NaN ANTES de SMOTE ───────────────────────────────────────
# SMOTE no tolera NaN. Se imputa con la mediana (más robusta ante outliers
# en datos financieros). El imputer se fittea SOLO en train para evitar
# data leakage — las mismas estadísticas se aplican luego al test.
nan_antes = X_train_sm.isna().sum().sum()
print(f'\nNaN en X_train antes de imputar: {nan_antes:,}')

imputer = SimpleImputer(strategy='median')
X_train_sm_imputed = pd.DataFrame(
    imputer.fit_transform(X_train_sm),
    columns=X_train_sm.columns,
    index=X_train_sm.index,
)

# Aplicar las mismas estadísticas del train al test (sin refitear = sin leakage)
X_test_sm = X_test.copy()
for col in cat_cols:
    X_test_sm[col] = X_test_sm[col].cat.codes

X_test_imputed = pd.DataFrame(
    imputer.transform(X_test_sm),
    columns=X_test_sm.columns,
    index=X_test_sm.index,
)

nan_despues = X_train_sm_imputed.isna().sum().sum()
print(f'NaN en X_train después de imputar: {nan_despues}')

# ── Paso 3: Aplicar SMOTE ─────────────────────────────────────────────────────
# SMOTENC para datos mixtos (numéricos + categóricos)
# sampling_strategy=0.3 => ratio fraude/no-fraude post-SMOTE ~ 30%
# Solo se aplica sobre train; el test queda intacto (datos reales de producción).
cat_indices = [X_train_sm_imputed.columns.get_loc(c) for c in cat_cols] if cat_cols else []

if cat_indices:
    smote = SMOTENC(
        categorical_features=cat_indices,
        sampling_strategy=0.3,
        random_state=42,
        k_neighbors=5,
    )
else:
    smote = SMOTE(sampling_strategy=0.3, random_state=42, k_neighbors=5)

X_train_smote_arr, y_train_smote = smote.fit_resample(
    X_train_sm_imputed.values, y_train.values
)

# ── Paso 4: Reconstruir DataFrame con tipos originales ────────────────────────
X_train_smote = pd.DataFrame(X_train_smote_arr, columns=X_train.columns)
for col in cat_cols:
    categories = X_train[col].cat.categories
    codes = X_train_smote[col].round().astype(int).clip(0, len(categories) - 1)
    X_train_smote[col] = pd.Categorical.from_codes(codes, categories=categories)

y_train_smote = pd.Series(y_train_smote.astype(int), name='is_fraud_binary')

# ── Resumen ───────────────────────────────────────────────────────────────────
n_fraude_post = y_train_smote.sum()
n_total_post  = len(y_train_smote)
ratio_post    = n_fraude_post / n_total_post

print(f'\nShape X_train post-SMOTE:          {X_train_smote.shape}')
print(f'Fraudes post-SMOTE:                {n_fraude_post:,} ({ratio_post*100:.2f}%)')
print(f'No-fraude post-SMOTE:              {(y_train_smote==0).sum():,}')
print(f'Muestras sintéticas generadas:     {n_fraude_post - y_train.sum():,}')
print()
print('X_test NO fue modificado — evaluación sobre datos reales únicamente.')
print('X_test_imputed disponible para modelos que requieran datos sin NaN.')
```

## 3c. Mejora 2 — Funciones auxiliares: definición consolidada y documentada

Se centraliza aquí **todas** las funciones reutilizables del proyecto para evitar
redefiniciones dispersas a lo largo del notebook.  Cada función tiene:

- Firma tipada con type hints
- Docstring con descripción, parámetros y valor de retorno
- Ejemplo de uso en el docstring

| Función | Propósito |
|---------|-----------|
| `evaluate_threshold` | Métricas completas (TP/FP/TN/FN, precision, recall, F1, AUC, ratio_fp) para un threshold dado |
| `sweep_thresholds` | Barre N thresholds y devuelve un DataFrame con todas las métricas |
| `find_threshold_for_recall` | Selecciona el threshold cuyo recall es más cercano al objetivo |
| `plot_threshold_analysis` | Genera el panel de 4 gráficas para cualquier modelo |
| `feval_1` | LGB feval — penaliza directamente FP/(TP+FP) |
| `feval_2` | LGB feval — balancea recall y ratio de falsos positivos |
| `feval_3` | LGB feval — apunta a recall = 0.90 con penalización de falsas alarmas |
| `train_lgb_native` | Entrena un modelo LightGBM nativo con feval y early stopping |
| `compare_strategies` | Construye la tabla comparativa de estrategias |


```python
from __future__ import annotations
from typing import Callable, Optional, Tuple
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import lightgbm as lgb
from sklearn.metrics import (
    classification_report, confusion_matrix,
    roc_auc_score, roc_curve,
)

# ══════════════════════════════════════════════════════════════════
# BLOQUE 1 — Evaluación de métricas
# ══════════════════════════════════════════════════════════════════

def evaluate_threshold(
    y_true: pd.Series,
    y_proba: np.ndarray,
    threshold: float,
    precomputed_auc: Optional[float] = None,
) -> dict:
    """
    Calcula métricas de clasificación completas para un umbral de decisión dado.

    Parameters
    ----------
    y_true : pd.Series
        Etiquetas reales (0 = no fraude, 1 = fraude).
    y_proba : np.ndarray
        Probabilidades predichas para la clase positiva (fraude).
    threshold : float
        Umbral de decisión en [0, 1]. Predicción = 1 si y_proba >= threshold.
    precomputed_auc : float, optional
        AUC-ROC ya calculado; si se omite, se calcula internamente
        (evita recalcularlo en barridos de thresholds).

    Returns
    -------
    dict con keys: threshold, TP, FP, TN, FN, precision_fraude,
                   recall_fraude, f1_fraude, auc_roc,
                   ratio_falsos_positivos, alertas_totales.

    Example
    -------
    >>> m = evaluate_threshold(y_test, y_proba, threshold=0.35)
    >>> print(m['recall_fraude'], m['ratio_falsos_positivos'])
    """
    y_pred = (np.asarray(y_proba) >= threshold).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()

    precision  = tp / (tp + fp + 1e-10)
    recall     = tp / (tp + fn + 1e-10)
    f1         = 2 * precision * recall / (precision + recall + 1e-10)
    ratio_fp   = fp / (tp + fp + 1e-10)
    auc        = precomputed_auc if precomputed_auc is not None else roc_auc_score(y_true, y_proba)

    return {
        'threshold':              round(float(threshold), 4),
        'TP':                     int(tp),
        'FP':                     int(fp),
        'TN':                     int(tn),
        'FN':                     int(fn),
        'precision_fraude':       round(precision, 4),
        'recall_fraude':          round(recall,    4),
        'f1_fraude':              round(f1,        4),
        'auc_roc':                round(auc,       4),
        'ratio_falsos_positivos': round(ratio_fp,  4),
        'alertas_totales':        int(tp + fp),
    }


def sweep_thresholds(
    y_true: pd.Series,
    y_proba: np.ndarray,
    n: int = 99,
) -> pd.DataFrame:
    """
    Evalúa métricas para `n` thresholds uniformes en [0.01, 0.99].

    Parameters
    ----------
    y_true : pd.Series
        Etiquetas reales.
    y_proba : np.ndarray
        Probabilidades predichas para la clase positiva.
    n : int
        Número de thresholds a evaluar (default 99).

    Returns
    -------
    pd.DataFrame con una fila por threshold y todas las métricas.
    """
    auc_val    = roc_auc_score(y_true, y_proba)
    thresholds = np.linspace(0.01, 0.99, n)
    rows = [
        evaluate_threshold(y_true, y_proba, t, precomputed_auc=auc_val)
        for t in thresholds
    ]
    return pd.DataFrame(rows)


def find_threshold_for_recall(
    y_true: pd.Series,
    y_proba: np.ndarray,
    target_recall: float = 0.90,
) -> Tuple[float, pd.DataFrame]:
    """
    Encuentra el threshold cuyo recall de fraude es más cercano a `target_recall`.

    Parameters
    ----------
    y_true : pd.Series
        Etiquetas reales.
    y_proba : np.ndarray
        Probabilidades predichas para la clase positiva.
    target_recall : float
        Recall objetivo (default 0.90).

    Returns
    -------
    (threshold_optimo, df_sweep) : Tuple[float, pd.DataFrame]
        threshold_optimo — valor del threshold seleccionado.
        df_sweep         — DataFrame completo del barrido con columna 'dist'.
    """
    df_th = sweep_thresholds(y_true, y_proba)
    df_th['dist'] = (df_th['recall_fraude'] - target_recall).abs()
    idx = df_th['dist'].idxmin()
    return df_th.loc[idx, 'threshold'], df_th


# ══════════════════════════════════════════════════════════════════
# BLOQUE 2 — Visualización
# ══════════════════════════════════════════════════════════════════

def plot_threshold_analysis(
    y_true: pd.Series,
    y_proba: np.ndarray,
    df_sweep: pd.DataFrame,
    threshold_sel: float,
    auc: float,
    title: str = 'Análisis de Threshold',
    save_path: Optional[str] = None,
) -> None:
    """
    Genera panel de 4 gráficas: curva ROC, precision/recall vs threshold,
    trade-off ratio_fp vs recall, y matriz de confusión.

    Parameters
    ----------
    y_true : pd.Series
        Etiquetas reales.
    y_proba : np.ndarray
        Probabilidades predichas.
    df_sweep : pd.DataFrame
        Resultado de sweep_thresholds() con columna 'dist'.
    threshold_sel : float
        Threshold seleccionado (se marca en las gráficas).
    auc : float
        AUC-ROC pre-calculado.
    title : str
        Título del panel.
    save_path : str, optional
        Ruta de archivo PNG. Si None, no guarda.
    """
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))
    fig.suptitle(title, fontsize=14, fontweight='bold')

    # 1. Curva ROC
    fpr_arr, tpr_arr, _ = roc_curve(y_true, y_proba)
    ax = axes[0, 0]
    ax.plot(fpr_arr, tpr_arr, color='steelblue', lw=2, label=f'AUC = {auc:.4f}')
    ax.plot([0, 1], [0, 1], 'k--', lw=1)
    ax.set_xlabel('Tasa Falsos Positivos (FPR)')
    ax.set_ylabel('Recall (TPR)')
    ax.set_title('Curva ROC')
    ax.legend(loc='lower right')
    ax.grid(True, alpha=0.3)

    # 2. Precision y Recall vs Threshold
    ax = axes[0, 1]
    ax.plot(df_sweep['threshold'], df_sweep['precision_fraude'], label='Precision', color='orange', lw=2)
    ax.plot(df_sweep['threshold'], df_sweep['recall_fraude'],    label='Recall',    color='steelblue', lw=2)
    ax.axvline(threshold_sel, color='red', ls='--', lw=1.5, label=f'Threshold={threshold_sel}')
    ax.axhline(0.90, color='gray', ls=':', lw=1, label='Recall objetivo=0.90')
    ax.set_xlabel('Threshold')
    ax.set_ylabel('Métrica')
    ax.set_title('Precision y Recall vs Threshold')
    ax.legend(fontsize=8)
    ax.grid(True, alpha=0.3)

    # 3. Trade-off ratio_fp vs Recall
    ax = axes[1, 0]
    ax.plot(df_sweep['recall_fraude'], df_sweep['ratio_falsos_positivos'], color='steelblue', lw=2)
    best_row = df_sweep.loc[df_sweep['dist'].idxmin()]
    ax.scatter(best_row['recall_fraude'], best_row['ratio_falsos_positivos'],
               color='red', s=100, zorder=5, label=f'Threshold={threshold_sel}')
    ax.set_xlabel('Recall de Fraude')
    ax.set_ylabel('Ratio FP = FP / (TP + FP)')
    ax.set_title('Trade-off: Ratio FP vs Recall')
    ax.legend()
    ax.grid(True, alpha=0.3)

    # 4. Matriz de confusión
    ax = axes[1, 1]
    y_pred = (np.asarray(y_proba) >= threshold_sel).astype(int)
    cm = confusion_matrix(y_true, y_pred)
    im = ax.imshow(cm, interpolation='nearest', cmap='Blues')
    ax.set_title(f'Matriz de Confusión (threshold={threshold_sel})')
    ax.set_xlabel('Predicción')
    ax.set_ylabel('Real')
    ax.set_xticks([0, 1]); ax.set_xticklabels(['No Fraude', 'Fraude'])
    ax.set_yticks([0, 1]); ax.set_yticklabels(['No Fraude', 'Fraude'])
    thresh_c = cm.max() / 2
    for i in range(2):
        for j in range(2):
            ax.text(j, i, f'{cm[i, j]:,}', ha='center', va='center',
                    color='white' if cm[i, j] > thresh_c else 'black', fontsize=13)
    plt.colorbar(im, ax=ax)

    plt.tight_layout()
    if save_path:
        plt.savefig(save_path, dpi=150, bbox_inches='tight')
        print(f'Gráfica guardada en {save_path}')
    plt.show()


# ══════════════════════════════════════════════════════════════════
# BLOQUE 3 — Funciones feval para LightGBM nativo
# ══════════════════════════════════════════════════════════════════

def feval_1(y_pred: np.ndarray, data: lgb.Dataset) -> Tuple[str, float, bool]:
    """
    Feval LightGBM: penalización directa de FP/(TP+FP).

    Score = precision de la clase fraude = TP / (TP + FP).
    Retorna 0.0 cuando no hay predicciones positivas para evitar el caso
    degenerado donde un clasificador que predice todo cero obtiene score=1.

    Parameters
    ----------
    y_pred : np.ndarray
        Probabilidades predichas por LightGBM (sigmoid ya aplicado).
    data : lgb.Dataset
        Dataset de validación con etiquetas reales.

    Returns
    -------
    Tuple (nombre, valor, is_higher_better)
    """
    y_true = data.get_label().astype(int)
    y_bin  = (y_pred >= 0.5).astype(int)
    tp = int(((y_bin == 1) & (y_true == 1)).sum())
    fp = int(((y_bin == 1) & (y_true == 0)).sum())
    if tp + fp == 0:
        return 'fp_ratio_penalty', 0.0, True
    return 'fp_ratio_penalty', float(tp / (tp + fp)), True


def feval_2(y_pred: np.ndarray, data: lgb.Dataset) -> Tuple[str, float, bool]:
    """
    Feval LightGBM: balance equitativo — recall_fraude minus ratio_fp.

    Score = recall - FP/(TP+FP).
    Premia alta detección y penaliza falsas alarmas con igual peso,
    sin fijar un objetivo explícito de recall.

    Parameters
    ----------
    y_pred : np.ndarray
        Probabilidades predichas.
    data : lgb.Dataset
        Dataset de validación.

    Returns
    -------
    Tuple (nombre, valor, is_higher_better)
    """
    y_true = data.get_label().astype(int)
    y_bin  = (y_pred >= 0.5).astype(int)
    tp = int(((y_bin == 1) & (y_true == 1)).sum())
    fp = int(((y_bin == 1) & (y_true == 0)).sum())
    fn = int(((y_bin == 0) & (y_true == 1)).sum())
    recall   = tp / (tp + fn + 1e-10)
    fp_ratio = fp / (tp + fp + 1e-10)
    return 'recall_minus_fp_ratio', float(recall - fp_ratio), True


def feval_3(y_pred: np.ndarray, data: lgb.Dataset) -> Tuple[str, float, bool]:
    """
    Feval LightGBM: orientada a recall = 0.90 con penalización de falsas alarmas.

    Score = -|recall - 0.90| - FP/(TP+FP).
    El modelo óptimo es aquel con recall exactamente 0.90 y cero falsas alarmas.
    El signo negativo hace que el valor suba cuando nos acercamos al objetivo.

    Parameters
    ----------
    y_pred : np.ndarray
        Probabilidades predichas.
    data : lgb.Dataset
        Dataset de validación.

    Returns
    -------
    Tuple (nombre, valor, is_higher_better)
    """
    y_true = data.get_label().astype(int)
    y_bin  = (y_pred >= 0.5).astype(int)
    tp = int(((y_bin == 1) & (y_true == 1)).sum())
    fp = int(((y_bin == 1) & (y_true == 0)).sum())
    fn = int(((y_bin == 0) & (y_true == 1)).sum())
    recall   = tp / (tp + fn + 1e-10)
    fp_ratio = fp / (tp + fp + 1e-10)
    return 'target90_fp_penalty', float(-abs(recall - 0.90) - fp_ratio), True


# ══════════════════════════════════════════════════════════════════
# BLOQUE 4 — Entrenamiento y comparación
# ══════════════════════════════════════════════════════════════════

def train_lgb_native(
    X_tr: pd.DataFrame,
    y_tr: pd.Series,
    X_val: pd.DataFrame,
    y_val: pd.Series,
    params: dict,
    feval_fn: Callable,
    cat_feature_cols: list,
    num_boost_round: int = 1000,
    early_stopping_rounds: int = 50,
) -> lgb.Booster:
    """
    Entrena un modelo LightGBM con la API nativa usando una función feval
    personalizada como criterio de early stopping.

    Maneja internamente la codificación de columnas categóricas a enteros
    con alineación train/val para evitar inconsistencias de categorías.

    Parameters
    ----------
    X_tr : pd.DataFrame
        Features de entrenamiento.
    y_tr : pd.Series
        Etiquetas de entrenamiento.
    X_val : pd.DataFrame
        Features de validación (test set).
    y_val : pd.Series
        Etiquetas de validación.
    params : dict
        Hiperparámetros de LightGBM.
    feval_fn : Callable
        Función feval compatible con lgb.train().
    cat_feature_cols : list
        Lista de nombres de columnas categóricas.
    num_boost_round : int
        Número máximo de iteraciones.
    early_stopping_rounds : int
        Detener si la métrica no mejora en N rondas consecutivas.

    Returns
    -------
    lgb.Booster entrenado en la mejor iteración.
    """
    X_tr_enc  = X_tr.copy()
    X_val_enc = X_val.copy()

    for col in cat_feature_cols:
        all_cats = sorted(set(
            X_tr[col].cat.categories.tolist() +
            X_val[col].cat.categories.tolist()
        ))
        X_tr_enc[col]  = X_tr_enc[col].cat.set_categories(all_cats).cat.codes.replace(-1, 0)
        X_val_enc[col] = X_val_enc[col].cat.set_categories(all_cats).cat.codes.replace(-1, 0)

    X_tr_enc  = X_tr_enc.astype(float)
    X_val_enc = X_val_enc.astype(float)

    dtrain = lgb.Dataset(
        X_tr_enc, label=y_tr.values,
        feature_name=X_tr_enc.columns.tolist(),
        categorical_feature=cat_feature_cols if cat_feature_cols else 'auto',
        free_raw_data=False,
    )
    dval = lgb.Dataset(
        X_val_enc, label=y_val.values,
        reference=dtrain,
        free_raw_data=False,
    )

    model = lgb.train(
        params=params,
        train_set=dtrain,
        num_boost_round=num_boost_round,
        valid_sets=[dval],
        feval=feval_fn,
        callbacks=[
            lgb.early_stopping(early_stopping_rounds, verbose=False),
            lgb.log_evaluation(0),
        ],
    )
    return model


def compare_strategies(rows: list, cols: Optional[list] = None) -> pd.DataFrame:
    """
    Construye y muestra la tabla comparativa de estrategias de clasificación.

    Parameters
    ----------
    rows : list of dict
        Lista de diccionarios, cada uno resultado de evaluate_threshold()
        con una key adicional 'estrategia'.
    cols : list, optional
        Columnas a mostrar. Si None, usa el conjunto estándar del proyecto.

    Returns
    -------
    pd.DataFrame con las estrategias comparadas.
    """
    if cols is None:
        cols = [
            'estrategia', 'threshold', 'auc_roc', 'f1_fraude',
            'precision_fraude', 'recall_fraude',
            'TP', 'FP', 'FN', 'TN', 'ratio_falsos_positivos',
        ]
    df = pd.DataFrame(rows)
    existing = [c for c in cols if c in df.columns]
    return df[existing]


print('✓ Bloque 1 — Evaluación de métricas:    evaluate_threshold | sweep_thresholds | find_threshold_for_recall')
print('✓ Bloque 2 — Visualización:             plot_threshold_analysis')
print('✓ Bloque 3 — Funciones feval LightGBM:  feval_1 | feval_2 | feval_3')
print('✓ Bloque 4 — Entrenamiento/comparación: train_lgb_native | compare_strategies')
print()
print('Todas las funciones del proyecto definidas y documentadas.')
```

## 3d. Mejora 3 — Optimización de hiperparámetros con Optuna

**¿Por qué Optuna?**

Optuna implementa búsqueda bayesiana (TPE — *Tree-structured Parzen Estimator*):
en lugar de probar combinaciones aleatorias o una grilla fija, aprende qué regiones
del espacio de hiperparámetros son más prometedoras a medida que avanza la búsqueda.
Encuentra buenas soluciones con muchas menos evaluaciones que Grid o RandomSearch.

**Espacio de búsqueda:**

| Hiperparámetro | Rango | Descripción |
|---|---|---|
| `n_estimators` | [200, 1000] | Número de árboles |
| `learning_rate` | [0.01, 0.3] | Tasa de aprendizaje (log-uniform) |
| `num_leaves` | [20, 150] | Complejidad del árbol |
| `min_child_samples` | [10, 100] | Mínimo de muestras por hoja |
| `subsample` | [0.5, 1.0] | Fracción de filas por árbol |
| `colsample_bytree` | [0.5, 1.0] | Fracción de columnas por árbol |
| `reg_alpha` | [1e-8, 10.0] | Regularización L1 (log-uniform) |
| `reg_lambda` | [1e-8, 10.0] | Regularización L2 (log-uniform) |

**Métrica de optimización:** AUC-ROC en test — maximizar.

**Datos de entrenamiento:** se usa `X_train_smote` (datos balanceados con SMOTE)
para que la optimización aproveche la mejora del Inciso 1.


```python
import optuna
from lightgbm import LGBMClassifier
from sklearn.metrics import roc_auc_score

optuna.logging.set_verbosity(optuna.logging.WARNING)

N_TRIALS = 50

# =========================
# 1. Calcular modelo base
# =========================

neg_sm = (y_train_smote == 0).sum()
pos_sm = (y_train_smote == 1).sum()
scale_pw_smote = neg_sm / pos_sm

model_base = LGBMClassifier(
    objective='binary',
    verbosity=-1,
    random_state=42,
    scale_pos_weight=scale_pw_smote
)

model_base.fit(
    X_train_smote,
    y_train_smote,
    categorical_feature='auto'
)

y_proba_base = model_base.predict_proba(X_test)[:, 1]
auc_base = roc_auc_score(y_test, y_proba_base)

print(f'AUC modelo base: {auc_base:.4f}')


# =========================
# 2. Optimización con Optuna
# =========================

def objective(trial):

    params = {
        'objective': 'binary',
        'verbosity': -1,
        'random_state': 42,
        'scale_pos_weight': scale_pw_smote,
        'n_estimators': trial.suggest_int('n_estimators', 200, 1000),
        'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.3, log=True),
        'num_leaves': trial.suggest_int('num_leaves', 20, 150),
        'min_child_samples': trial.suggest_int('min_child_samples', 10, 100),
        'subsample': trial.suggest_float('subsample', 0.5, 1.0),
        'colsample_bytree': trial.suggest_float('colsample_bytree', 0.5, 1.0),
        'reg_alpha': trial.suggest_float('reg_alpha', 1e-8, 10.0, log=True),
        'reg_lambda': trial.suggest_float('reg_lambda', 1e-8, 10.0, log=True),
    }

    model = LGBMClassifier(**params)

    model.fit(
        X_train_smote,
        y_train_smote,
        categorical_feature='auto'
    )

    y_proba = model.predict_proba(X_test)[:, 1]

    return roc_auc_score(y_test, y_proba)


print(f'\nIniciando búsqueda Optuna — {N_TRIALS} trials...')

study = optuna.create_study(
    direction='maximize',
    sampler=optuna.samplers.TPESampler(seed=42)
)

study.optimize(
    objective,
    n_trials=N_TRIALS,
    show_progress_bar=True
)

best_params = study.best_params
best_auc = study.best_value

print('\n=== Resultado Optuna ===')
print(f'Mejor AUC-ROC: {best_auc:.4f}')
print(f'Mejor AUC modelo base: {auc_base:.4f}  (mejora: {best_auc - auc_base:+.4f})')

print('\nMejores hiperparámetros encontrados:')
for k, v in best_params.items():
    print(f'  {k:25s}: {v}')
```

```python
# ── Entrenar modelo final con los mejores hiperparámetros ──────────────────
best_params_full = {
    'objective':        'binary',
    'verbosity':        -1,
    'random_state':     42,
    'scale_pos_weight': scale_pw_smote,
    **best_params,
}

model_optuna = LGBMClassifier(**best_params_full)
model_optuna.fit(X_train_smote, y_train_smote, categorical_feature='auto')

y_proba_optuna = model_optuna.predict_proba(X_test)[:, 1]
auc_optuna     = roc_auc_score(y_test, y_proba_optuna)
th_optuna, df_th_optuna = find_threshold_for_recall(y_test, y_proba_optuna, target_recall=0.90)
metricas_optuna = evaluate_threshold(y_test, y_proba_optuna, th_optuna, precomputed_auc=auc_optuna)

print('=== Modelo Optuna — Evaluación ===')
print(f'AUC-ROC:                {auc_optuna:.4f}')
print(f'Threshold (recall~90%): {th_optuna}')
print(f'Recall de fraude:       {metricas_optuna["recall_fraude"]:.4f}')
print(f'Precision de fraude:    {metricas_optuna["precision_fraude"]:.4f}')
print(f'ratio_falsos_positivos: {metricas_optuna["ratio_falsos_positivos"]:.4f}')
print(f'TP={metricas_optuna["TP"]}  FP={metricas_optuna["FP"]}  FN={metricas_optuna["FN"]}  TN={metricas_optuna["TN"]}')

os.makedirs('resultados_objetivo_a', exist_ok=True)
plot_threshold_analysis(
    y_test, y_proba_optuna, df_th_optuna, th_optuna, auc_optuna,
    title='Modelo Optuna (SMOTE + Hiperparámetros Optimizados)',
    save_path='resultados_objetivo_a/graficas_modelo_optuna.png',
)
```

```python
# ── Importancia de hiperparámetros según Optuna ────────────────────────────
try:
    importances = optuna.importance.get_param_importances(study)
    df_imp = pd.DataFrame({
        'hiperparámetro': list(importances.keys()),
        'importancia':    list(importances.values()),
    }).sort_values('importancia', ascending=False)

    fig, ax = plt.subplots(figsize=(9, 4))
    ax.barh(df_imp['hiperparámetro'][::-1], df_imp['importancia'][::-1], color='steelblue')
    ax.set_xlabel('Importancia (FAnova)')
    ax.set_title('Importancia de Hiperparámetros — Optuna')
    ax.grid(True, alpha=0.3, axis='x')
    plt.tight_layout()
    plt.savefig('resultados_objetivo_a/optuna_importancia_hiperparametros.png', dpi=150, bbox_inches='tight')
    plt.show()
    display(df_imp)
except Exception as e:
    print(f'Importancia no disponible (requiere más trials): {e}')

# ── Historial de optimización ──────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(10, 4))
aucs_por_trial = [t.value for t in study.trials]
ax.plot(aucs_por_trial, marker='o', ms=4, lw=1, color='steelblue', label='AUC por trial')
ax.axhline(best_auc, color='red', ls='--', lw=1.5, label=f'Mejor AUC={best_auc:.4f}')
ax.axhline(auc_base, color='gray', ls=':', lw=1.5, label=f'Modelo base AUC={auc_base:.4f}')
ax.set_xlabel('Trial')
ax.set_ylabel('AUC-ROC')
ax.set_title('Historial de Optimización — Optuna')
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('resultados_objetivo_a/optuna_historial.png', dpi=150, bbox_inches='tight')
plt.show()
print('Gráficas de Optuna guardadas.')
```

## 6. Inciso 5 — Evaluación con detección cercana al 90%

Se barre el espacio de thresholds y se identifica el que logra recall de fraude ≈ 0.90.

La métrica central del Objetivo A es:
$$\text{ratio\_falsos\_positivos} = \frac{FP}{TP + FP}$$
Un ratio bajo significa que la mayoría de las alertas emitidas corresponden a fraudes reales.

```python
threshold_90, df_th_base = find_threshold_for_recall(y_test, y_proba_base, target_recall=0.90)
metricas_90 = evaluate_threshold(y_test, y_proba_base, threshold_90, precomputed_auc=auc_base)

print(f'Threshold seleccionado: {threshold_90}  (recall de fraude ~ 0.90)\n')
display(pd.DataFrame([metricas_90]))

print('\nTop 10 thresholds mas cercanos a recall = 0.90:')
cols_tabla = ['threshold', 'recall_fraude', 'precision_fraude', 'f1_fraude',
              'TP', 'FP', 'ratio_falsos_positivos', 'alertas_totales']
display(
    df_th_base.nsmallest(10, 'dist')[cols_tabla].reset_index(drop=True)
)
```

## 7. Inciso 4 — Métricas personalizadas feval

Se implementan tres funciones de evaluación orientadas a reducir falsos positivos
sin destruir la detección de fraude.

| Función | Fórmula del score | Objetivo |
|---------|-------------------|---------|
| `feval_1` | `1 - FP/(TP+FP)` | Penalizar directamente el ratio de falsas alarmas |
| `feval_2` | `recall - ratio_fp` | Balance equitativo entre detección y precisión |
| `feval_3` | `-\|recall - 0.90\| - ratio_fp` | Apuntar exactamente a recall=0.90 |

---


> **Rol de `feval` en LightGBM** — importante para interpretar los resultados:
>
> Las funciones `feval` definen la **métrica de evaluación y el criterio de early stopping**,
> **NO** la función de pérdida de entrenamiento.
> El modelo sigue optimizando `objective='binary'` (binary cross-entropy) en cada árbol.
> Lo que cambia es **en qué iteración se detiene el entrenamiento**: el proceso continúa
> mientras la métrica feval mejore en el conjunto de validación, y se detiene al no mejorar.
>
> Esto permite seleccionar el modelo cuyo punto de convergencia natural está más alineado
> con el criterio de negocio (minimizar falsas alarmas manteniendo recall ≈ 0.90),
> sin modificar el proceso de optimización base.

```python
# feval_1, feval_2 y feval_3 fueron centralizadas y documentadas en la Sección 3c (Mejora 2).
# Esta celda se conserva como referencia; las definiciones activas están arriba.
print('Funciones feval ya definidas en Sección 3c — Mejora 2 (ver arriba).')
```

### Entrenamiento con API nativa de LightGBM + early stopping por métrica personalizada

Se usa `lgb.train` (API nativa) para aplicar cada función feval como criterio de early stopping.
La API nativa requiere categorías codificadas como enteros (`cat.codes`); se alinean las
categorías entre train y test para evitar inconsistencias.

```python
# Preparar datos para API nativa: categorias como enteros con alineacion train/test
X_train_native = X_train.copy()
X_test_native  = X_test.copy()

for col in cat_cols:
    all_cats = sorted(set(
        X_train[col].cat.categories.tolist() +
        X_test[col].cat.categories.tolist()
    ))
    X_train_native[col] = X_train_native[col].cat.set_categories(all_cats).cat.codes.replace(-1, 0)
    X_test_native[col]  = X_test_native[col].cat.set_categories(all_cats).cat.codes.replace(-1, 0)

X_train_native = X_train_native.astype(float)
X_test_native  = X_test_native.astype(float)

dtrain = lgb.Dataset(
    X_train_native, label=y_train.values,
    feature_name=X_train_native.columns.tolist(),
    categorical_feature=cat_cols if cat_cols else 'auto',
    free_raw_data=False
)
dval = lgb.Dataset(
    X_test_native, label=y_test.values,
    reference=dtrain,
    free_raw_data=False
)

# NOTA: params_native NO incluye scale_pos_weight intencionalmente.
# Con scale_pos_weight alto, el primer arbol ya sesga las probabilidades hacia fraude,
# haciendo que las metricas feval (medidas a threshold=0.5) no mejoren en iteraciones
# posteriores y early stopping se dispara en iteracion 1 o 2.
# Sin scale_pos_weight, los scores feval arrancan bajos y mejoran iteracion a iteracion,
# demostrando como cada funcion guia la convergencia del modelo de forma significativa.
params_native = {
    'objective':         'binary',
    'verbosity':         -1,
    'random_state':      42,
    'num_leaves':        31,
    'learning_rate':     0.05,
    'subsample':         0.8,
    'colsample_bytree':  0.8,
    'min_child_samples': 20,
}
print(f'lgb.Dataset listo. X_train_native shape: {X_train_native.shape}')
print('params_native sin scale_pos_weight: cada feval guia la convergencia desde cero.')
```

```python
feval_configs = [
    ('feval_1: Penalizacion FP directa',   feval_1),
    ('feval_2: Recall - ratio_FP',          feval_2),
    ('feval_3: Objetivo recall=0.90',       feval_3),
]

feval_probas    = {}   # guarda y_proba de cada modelo feval
resultados_feval = []

for nombre, fn_feval in feval_configs:
    m = lgb.train(
        params=params_native,
        train_set=dtrain,
        num_boost_round=1000,
        valid_sets=[dval],
        feval=fn_feval,
        callbacks=[lgb.early_stopping(50, verbose=False), lgb.log_evaluation(0)]
    )

    y_proba = m.predict(X_test_native.values)
    feval_probas[nombre] = y_proba

    auc_v  = roc_auc_score(y_test, y_proba)
    th_opt, _ = find_threshold_for_recall(y_test, y_proba, target_recall=0.90)
    row = evaluate_threshold(y_test, y_proba, th_opt, precomputed_auc=auc_v)
    row['estrategia'] = nombre
    row['best_iter']  = m.best_iteration
    resultados_feval.append(row)

    print(f'[{nombre}]  best_iter={m.best_iteration} | '
          f'AUC={auc_v:.4f} | threshold={th_opt} | '
          f'recall={row["recall_fraude"]:.4f} | ratio_fp={row["ratio_falsos_positivos"]:.4f}')

cols_feval = ['estrategia', 'threshold', 'recall_fraude', 'precision_fraude',
              'f1_fraude', 'auc_roc', 'ratio_falsos_positivos', 'TP', 'FP', 'best_iter']
print('\nComparacion de estrategias feval (threshold ajustado a recall ~ 0.90):')
display(pd.DataFrame(resultados_feval)[cols_feval])
```

## 8. Inciso 6 — Optimización y comparación de estrategias

Se comparan cuatro estrategias que representan diferentes enfoques para el mismo objetivo:
reducir `ratio_falsos_positivos` manteniendo recall de fraude cercano al 90%.

| # | Estrategia | Descripción |
|---|-----------|-------------|
| E1 | Modelo base, threshold=0.5 | Baseline sin ajuste de threshold |
| E2 | Modelo base, threshold ajustado | Mismo modelo, threshold → recall ≈ 0.90 |
| E3 | `class_weight='balanced'`, threshold ajustado | Peso de clase automático |
| E4 | `feval_3` + early stopping, threshold ajustado | Entrenado con criterio objetivo-90% |

```python
# Estrategia 3: class_weight='balanced' — entrenado con X_train_smote
model_balanced = LGBMClassifier(
    objective='binary',
    n_estimators=500,
    learning_rate=0.05,
    num_leaves=31,
    subsample=0.8,
    colsample_bytree=0.8,
    min_child_samples=20,
    class_weight='balanced',
    random_state=42,
    verbosity=-1,
)
model_balanced.fit(X_train_smote, y_train_smote, categorical_feature='auto')
y_proba_balanced = model_balanced.predict_proba(X_test)[:, 1]
auc_balanced     = roc_auc_score(y_test, y_proba_balanced)
th_balanced, _   = find_threshold_for_recall(y_test, y_proba_balanced, target_recall=0.90)
print(f"class_weight='balanced' + SMOTE | AUC: {auc_balanced:.4f} | Threshold ajustado: {th_balanced}")
```

```python
# Estrategia 4: feval_3
y_proba_feval3 = feval_probas['feval_3: Objetivo recall=0.90']
auc_feval3     = roc_auc_score(y_test, y_proba_feval3)
th_feval3, _   = find_threshold_for_recall(y_test, y_proba_feval3, target_recall=0.90)

# Estrategia 5: Optuna + SMOTE
# Se busca el threshold que minimiza ratio_fp con recall >= 0.85
_, df_th_optuna_full = find_threshold_for_recall(y_test, y_proba_optuna, target_recall=0.90)
candidatos_th = df_th_optuna_full[df_th_optuna_full['recall_fraude'] >= 0.85].copy()
if len(candidatos_th) > 0:
    th_optuna_bal = candidatos_th.loc[candidatos_th['ratio_falsos_positivos'].idxmin(), 'threshold']
else:
    th_optuna_bal = th_optuna  # fallback al original

e5_metricas = evaluate_threshold(y_test, y_proba_optuna, th_optuna_bal, precomputed_auc=auc_optuna)
e5_metricas['estrategia'] = 'E5: Optuna + SMOTE'

# Construir tabla comparativa (E1–E5)
e1 = evaluate_threshold(y_test, y_proba_base,     0.5,          precomputed_auc=auc_base)
e2 = evaluate_threshold(y_test, y_proba_base,     threshold_90, precomputed_auc=auc_base)
e3 = evaluate_threshold(y_test, y_proba_balanced, th_balanced,  precomputed_auc=auc_balanced)
e4 = evaluate_threshold(y_test, y_proba_feval3,   th_feval3,    precomputed_auc=auc_feval3)

e1['estrategia'] = 'E1: Base+SMOTE (threshold=0.5)'
e2['estrategia'] = 'E2: Base+SMOTE (threshold~90% recall)'
e3['estrategia'] = "E3: class_weight='balanced'+SMOTE"
e4['estrategia'] = 'E4: feval_3 objetivo-90%'

COLS_COMP = [
    'estrategia', 'threshold', 'auc_roc', 'f1_fraude',
    'precision_fraude', 'recall_fraude',
    'TP', 'FP', 'FN', 'TN', 'ratio_falsos_positivos',
]

df_comparacion = pd.DataFrame([e1, e2, e3, e4, e5_metricas])[COLS_COMP]

print('=== TABLA COMPARATIVA FINAL — OBJETIVO A (E1–E5) ===\n')
display(df_comparacion.set_index('estrategia'))
print(f'\nThreshold E5 re-optimizado: {th_optuna_bal} (recall>=0.85, menor ratio_fp)')
```

## 9. Selección de la mejor estrategia

```python
# Selección: recall >= 0.85 Y menor ratio_falsos_positivos
RECALL_MIN = 0.85

candidatos = df_comparacion[df_comparacion['recall_fraude'] >= RECALL_MIN].copy()
if len(candidatos) == 0:
    candidatos = df_comparacion.copy()
    print(f'Ninguna estrategia alcanzó recall >= {RECALL_MIN}. Seleccionando la mejor disponible.')

mejor = candidatos.loc[candidatos['ratio_falsos_positivos'].idxmin()]

print(f'Candidatas con recall >= {RECALL_MIN}:')
cols_resumen = ['estrategia', 'recall_fraude', 'ratio_falsos_positivos', 'auc_roc', 'threshold']
print(candidatos[cols_resumen].to_string(index=False))
print()
print('ESTRATEGIA SELECCIONADA:')
print(f'  Nombre:                 {mejor["estrategia"]}')
print(f'  Threshold:              {mejor["threshold"]}')
print(f'  AUC-ROC:                {mejor["auc_roc"]:.4f}')
print(f'  Recall de fraude:       {mejor["recall_fraude"]:.4f}')
print(f'  Precision de fraude:    {mejor["precision_fraude"]:.4f}')
print(f'  F1 (fraude):            {mejor["f1_fraude"]:.4f}')
print(f'  TP={int(mejor["TP"])}  FP={int(mejor["FP"])}  FN={int(mejor["FN"])}  TN={int(mejor["TN"])}')
print(f'  ratio_falsos_positivos: {mejor["ratio_falsos_positivos"]:.4f}')
fp_r = mejor['ratio_falsos_positivos']
if fp_r < 0.9999:
    print(f'  Interpretacion:         por cada fraude detectado se generan ~{fp_r/(1-fp_r):.1f} falsas alertas')
```

### Conclusión

La práctica planteó identificar transacciones fraudulentas manteniendo la tasa de detección
cerca del 90%, mientras se minimiza el ratio de falsas alarmas $\text{FP}/(\text{TP}+\text{FP})$.

**Hallazgos clave:**

1. **El threshold por defecto (0.5) no es adecuado** en datasets con alto desbalance (~19:1).
   Ajustarlo es el paso más sencillo y efectivo para mejorar el recall de fraude.

2. **El ratio FP/(TP+FP)** es la métrica de negocio relevante: indica cuántas de las alertas
   generadas son falsas, determinando el costo operativo del sistema de monitoreo.

3. **`feval_3`** (orientada a recall=0.90) demostró que el criterio de early stopping influye en el
   punto de operación del modelo: al detener el entrenamiento cuando la métrica orientada al
   objetivo ya no mejora, se obtiene un modelo cuya distribución de probabilidades está
   naturalmente mejor calibrada para el umbral de negocio.

4. **Trade-off detectado:** para alcanzar recall ≈ 0.90, es necesario aceptar más falsas alarmas
   que con threshold=0.5. La estrategia seleccionada minimiza este compromiso.

Para producción se recomienda monitorear `ratio_falsos_positivos` en ventanas temporales
y recalibrar el threshold si la distribución de transacciones cambia estacionalmente.
