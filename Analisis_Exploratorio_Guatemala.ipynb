```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', 100)
```

```python
df = pd.read_csv("Copia de Datasets 3 bancos\Copia de 03_gt_estatal_seed3_n100000.csv", sep=';')
```

```python
print("Dimensiones del dataset:")
print(df.shape)
```

Valores Nulos

```python
nulls = df.isnull().sum().sort_values(ascending=False)

display(nulls[nulls > 0])
```

Estadisticas Descriptivas

```python
df.describe()
```

Transacciones Aprobadas

```python
approved_counts = df['approved'].value_counts()

plt.figure(figsize=(5,5))
approved_counts.plot(kind='bar')

plt.title("Transacciones Aprobadas")
plt.xlabel("Approved")
plt.ylabel("Cantidad")

plt.show()
```

Distribucion de montos

```python
plt.figure(figsize=(8,5))

df['amount_usd'].hist(bins=50)

plt.title("Distribución de Montos en USD")
plt.xlabel("Monto USD")
plt.ylabel("Frecuencia")

plt.show()
```

Transacciones Internacionales

```python
intl_fraud = pd.crosstab(df['is_international'], df['is_fraud'])

display(intl_fraud)

intl_fraud.plot(kind='bar', figsize=(6,5))

plt.title("Fraude en Transacciones Internacionales")
plt.xlabel("Internacional")
plt.ylabel("Cantidad")

plt.show()
```

Baseline vs monto de transaccion

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df['client_baseline_amount'],
    df['amount_usd'],
    c=df['is_fraud'].astype(int),
    alpha=0.5
)

plt.title("Monto Transacción vs Baseline Cliente")
plt.xlabel("Baseline Cliente")
plt.ylabel("Monto USD")

plt.show()
```

Porcentaje de fraude

```python
print("Resumen rápido:")

print(f"Total transacciones: {len(df)}")
print(f"Promedio monto USD: {df['amount_usd'].mean():.2f}")
print(f"Máximo monto USD: {df['amount_usd'].max():.2f}")
print(f"Transacciones internacionales: {df['is_international'].sum()}")
```
