## Limpieza y preparación de datos

### Copia de trabajo

Creamos `df_clean = df.copy()` para no alterar el EDA anterior. Si algo sale mal durante la limpieza podemos volver al `df` original sin recargar el notebook.

```python
df_clean = df.copy()

print("Dimensiones originales:", df.shape)
print("Dimensiones copia de limpieza:", df_clean.shape)
```

### Normalizar nombres de columnas

Algunos CSV traen columnas con espacios al inicio/final (e.g. `" amount_usd"`). Eso causa errores invisibles al seleccionar columnas.

```python
df_clean.columns = df_clean.columns.str.strip()

print("Cantidad de columnas:", len(df_clean.columns))
print(df_clean.columns.tolist())

columnas_clave = [
    "channel",
    "amount_usd",
    "is_international",
    "distance_from_home_km",
    "hour_local",
    "DE18_merchant_category_code",
    "DE22_pos_entry_mode",
    "client_baseline_amount",
    "approved",
    "is_fraud"
]

faltantes = [col for col in columnas_clave if col not in df_clean.columns]
print("Columnas clave faltantes:", faltantes)
```

### Revisar duplicados

Los duplicados exactos pueden inflar patrones y hacer que el modelo aprenda señales artificiales. Solo eliminamos filas idénticas en **todas** sus columnas; no basándonos en campos individuales como `client_id` o `amount_usd`, ya que pueden existir transacciones legítimamente similares.

```python
duplicados = df_clean.duplicated().sum()
porcentaje_duplicados = duplicados / len(df_clean) * 100

print("Filas duplicadas:", duplicados)
print(f"Porcentaje de duplicados: {porcentaje_duplicados:.4f}%")

df_clean = df_clean.drop_duplicates().reset_index(drop=True)

print("Dimensiones después de eliminar duplicados exactos:", df_clean.shape)
```

### Revisar nulos

Antes de imputar necesitamos saber qué columnas tienen datos faltantes y cuánto representan. Prestamos especial atención a las columnas críticas para el modelo.

```python
tabla_nulos = pd.DataFrame({
    "nulos": df_clean.isnull().sum(),
    "porcentaje_nulos": (df_clean.isnull().mean() * 100).round(2)
}).sort_values("porcentaje_nulos", ascending=False)

print("Columnas con nulos:")
display(tabla_nulos[tabla_nulos["nulos"] > 0])

columnas_criticas = [
    "amount_usd",
    "client_baseline_amount",
    "distance_from_home_km",
    "hour_local",
    "channel",
    "approved",
    "is_fraud"
]

print("\nNulos en columnas críticas:")
display(tabla_nulos.loc[columnas_criticas])
```

### Eliminar columnas completamente vacías

Columnas con 100% de nulos no aportan información al modelo y solo ocupan espacio. Las eliminamos de `df_clean` sin tocar `df` original.

```python
columnas_100_nulos = tabla_nulos[tabla_nulos["porcentaje_nulos"] == 100].index.tolist()

print("Columnas con 100% de nulos:")
print(columnas_100_nulos)

df_clean = df_clean.drop(columns=columnas_100_nulos)

print("\nDimensiones después de eliminar columnas vacías:", df_clean.shape)
```

### Validar columna objetivo `is_fraud`

LightGBM necesita la variable objetivo limpia. Para clasificación binaria conviene 0/1. Conservamos `is_fraud` original y creamos `is_fraud_binary` para el modelo, manteniendo trazabilidad.

```python
print("Distribución de is_fraud:")
print(df_clean["is_fraud"].value_counts(dropna=False))
print("\nTipo:", df_clean["is_fraud"].dtype)

df_clean["is_fraud_binary"] = df_clean["is_fraud"].astype(int)

print("\nVerificación is_fraud vs is_fraud_binary:")
display(df_clean[["is_fraud", "is_fraud_binary"]].head())
```

### Conversión de tipos numéricos

A veces Pandas lee números como texto si el CSV trae separadores, espacios o valores raros. Forzamos la conversión con `errors="coerce"` para convertir valores inválidos en `NaN` en lugar de lanzar errores.

```python
columnas_numericas_clave = [
    "amount_usd",
    "distance_from_home_km",
    "hour_local",
    "client_baseline_amount"
]

for col in columnas_numericas_clave:
    df_clean[col] = pd.to_numeric(df_clean[col], errors="coerce")

print("Tipos después de conversión:")
print(df_clean[columnas_numericas_clave].dtypes)
```

### Limpieza de variables categóricas

Estandarizamos para evitar categorías duplicadas por espacios o mayúsculas/minúsculas. Por ejemplo `ECOM`, `ecom` y `ecom ` deben ser la misma categoría. **No agrupamos** categorías pequeñas en "OTROS" todavía; eso se hace en ingeniería de variables.

```python
columnas_categoricas_clave = [
    "channel",
    "DE18_merchant_category_code",
    "DE22_pos_entry_mode",
    "approved"
]

for col in columnas_categoricas_clave:
    df_clean[col] = (
        df_clean[col]
        .where(df_clean[col].notna(), np.nan)
        .astype("string")
        .str.strip()
        .str.upper()
    )
    df_clean[col] = df_clean[col].fillna("DESCONOCIDO")

for col in columnas_categoricas_clave:
    print(f"\n{col}")
    print(df_clean[col].value_counts(dropna=False).head(10))
```

### Validación de rangos y valores raros

Detectamos valores imposibles que podrían afectar el modelo: montos negativos, horas fuera de rango (0–23) o distancias negativas. Primero los contamos y luego inspeccionamos antes de actuar.

```python
validaciones = {
    "amount_usd_negativos": (df_clean["amount_usd"] < 0).sum(),
    "client_baseline_amount_negativos": (df_clean["client_baseline_amount"] < 0).sum(),
    "distance_from_home_km_negativos": (df_clean["distance_from_home_km"] < 0).sum(),
    "hour_local_fuera_rango": (~df_clean["hour_local"].between(0, 23)).sum()
}

print("Valores fuera de rango:")
for k, v in validaciones.items():
    print(f"  {k}: {v}")

if (df_clean["amount_usd"] < 0).any():
    print("\nEjemplo de filas con amount_usd negativo:")
    display(df_clean[df_clean["amount_usd"] < 0].head())
```

### Tratamiento básico de valores imposibles

Convertimos valores imposibles claros a `NaN` sin eliminar filas, para no perder información del resto de columnas. No aplicamos capping de outliers todavía — eso se hace en la etapa de ingeniería de variables.

```python
df_clean.loc[df_clean["amount_usd"] < 0, "amount_usd"] = np.nan
df_clean.loc[df_clean["client_baseline_amount"] < 0, "client_baseline_amount"] = np.nan
df_clean.loc[df_clean["distance_from_home_km"] < 0, "distance_from_home_km"] = np.nan
df_clean.loc[~df_clean["hour_local"].between(0, 23), "hour_local"] = np.nan

print("Nulos en columnas numéricas clave tras tratamiento:")
print(df_clean[columnas_numericas_clave].isnull().sum())
```

### Imputación simple

Antes de imputar `distance_from_home_km`, creamos una bandera binaria que indica si el valor estaba ausente originalmente. En detección de fraude, la **ausencia de un dato también puede ser señal**. Usamos mediana para las variables numéricas (robusta a outliers). Las categóricas ya fueron imputadas con `"DESCONOCIDO"` en el paso 8.7. **No imputamos `is_fraud`**.

```python
# Bandera de nulos antes de imputar
df_clean["distance_from_home_km_was_missing"] = df_clean["distance_from_home_km"].isna().astype(int)
print(f"Transacciones con distance_from_home_km faltante: {df_clean['distance_from_home_km_was_missing'].sum()}")

for col in columnas_numericas_clave:
    mediana = df_clean[col].median()
    df_clean[col] = df_clean[col].fillna(mediana)
    print(f"{col}: nulos imputados con mediana = {mediana:.4f}")

print("\nNulos restantes en columnas numéricas clave:")
print(df_clean[columnas_numericas_clave].isnull().sum())
```

### Variables derivadas básicas

Creamos variables que capturan desviaciones del comportamiento normal del cliente, tal como lo describe el documento de PlusTI. Agregamos además **banderas de outliers** — en fraude, los montos extremos no deben eliminarse sino marcarse, porque pueden ser precisamente la señal que busca el modelo.

- `amount_diff_baseline` / `amount_vs_baseline`: qué tanto se aleja el monto del baseline.
- `is_night_txn`: transacción en horario 0–5 h.
- `high_distance_from_home`: distancia en percentil 95+.
- `high_amount_p95` / `high_amount_p99`: monto en percentil 95+/99+.
- `high_amount_vs_baseline`: ratio monto/baseline en percentil 95+.

```python
df_clean["amount_diff_baseline"] = (
    df_clean["amount_usd"] - df_clean["client_baseline_amount"]
)

df_clean["amount_vs_baseline"] = (
    df_clean["amount_usd"] / df_clean["client_baseline_amount"]
)

df_clean["is_night_txn"] = df_clean["hour_local"].between(0, 5).astype(int)

df_clean["high_distance_from_home"] = (
    df_clean["distance_from_home_km"] > df_clean["distance_from_home_km"].quantile(0.95)
).astype(int)

# Tratar divisiones por cero
df_clean.replace([np.inf, -np.inf], np.nan, inplace=True)
df_clean["amount_vs_baseline"] = df_clean["amount_vs_baseline"].fillna(
    df_clean["amount_vs_baseline"].median()
)

# Banderas de valores extremos
p95_amount = df_clean["amount_usd"].quantile(0.95)
p99_amount = df_clean["amount_usd"].quantile(0.99)

df_clean["high_amount_p95"] = (df_clean["amount_usd"] > p95_amount).astype(int)
df_clean["high_amount_p99"] = (df_clean["amount_usd"] > p99_amount).astype(int)

p95_ratio = df_clean["amount_vs_baseline"].quantile(0.95)
df_clean["high_amount_vs_baseline"] = (
    df_clean["amount_vs_baseline"] > p95_ratio
).astype(int)

print("Variables derivadas creadas:")
display(df_clean[[
    "amount_usd", "client_baseline_amount",
    "amount_diff_baseline", "amount_vs_baseline",
    "is_night_txn", "high_distance_from_home",
    "high_amount_p95", "high_amount_p99", "high_amount_vs_baseline"
]].describe())
```

### Validación final del dataset limpio

Verificamos que la limpieza no dañó el dataset: dimensiones, nulos restantes en columnas clave, y que la distribución de fraude se mantiene respecto al dataset original.

```python
columnas_validacion = columnas_clave + ["is_fraud_binary"]

print("Dimensiones finales:", df_clean.shape)

print("\nNulos restantes en columnas clave:")
print(df_clean[columnas_validacion].isnull().sum())

print("\nDistribución de is_fraud:")
print(df_clean["is_fraud"].value_counts(dropna=False))

print("\nDistribución de is_fraud_binary:")
print(df_clean["is_fraud_binary"].value_counts(dropna=False))

print(f"\nPorcentaje de fraude en df_clean: {df_clean['is_fraud_binary'].mean()*100:.2f}%")
print(f"Porcentaje de fraude en df original:  {df['is_fraud'].mean()*100:.2f}%")
```

### Creación de dataset listo para modelado

Creamos `df_model_ready` como versión más estricta de `df_clean`. Excluimos: la columna con >95% de nulos (`DE44_additional_response_data`), identificadores directos de alta cardinalidad que podrían causar memorización, y columnas casi vacías. `df_clean` se conserva íntegro para trazabilidad y para crear variables agregadas por cliente en la siguiente etapa.

```python
df_model_ready = df_clean.copy()

print("Dimensiones df_clean:", df_clean.shape)
print("Dimensiones df_model_ready inicial:", df_model_ready.shape)

# Excluir columnas con más de 95% de nulos
columnas_alto_nulo = tabla_nulos[
    (tabla_nulos["porcentaje_nulos"] > 95) &
    (tabla_nulos["porcentaje_nulos"] < 100)
].index.tolist()

print("\nColumnas con más de 95% de nulos (excluidas del modelo):")
print(columnas_alto_nulo)

df_model_ready = df_model_ready.drop(columns=columnas_alto_nulo, errors="ignore")

# Excluir identificadores directos de alta cardinalidad
columnas_identificadoras = [
    "transaction_id",
    "client_id",
    "pan_masked",
    "pan_hash",
    "DE2_PAN",
    "DE35_track2_data_masked",
    "DE37_retrieval_reference_number",
    "DE38_authorization_code",
    "DE41_terminal_id",
    "DE42_card_acceptor_id",
    "DE102_account_id_1",
    "DE103_account_id_2"
]

columnas_identificadoras_existentes = [
    col for col in columnas_identificadoras if col in df_model_ready.columns
]

df_model_ready = df_model_ready.drop(columns=columnas_identificadoras_existentes)

print("\nColumnas identificadoras eliminadas:")
print(columnas_identificadoras_existentes)

print("\nDimensiones después de eliminar identificadores:", df_model_ready.shape)
```

### Revisión de columnas constantes

Si una columna tiene un solo valor para todas las filas no aporta información discriminativa. En Bolivia, columnas como `bank_code`, `bank_name`, `bank_country` o `MTI` probablemente sean constantes al ser un dataset de un solo banco. El código las detecta sin asumirlo manualmente.

```python
nunique = df_model_ready.nunique(dropna=False).sort_values()

columnas_constantes = nunique[nunique <= 1].index.tolist()

print("Columnas constantes o con un solo valor:")
print(columnas_constantes)

df_model_ready = df_model_ready.drop(columns=columnas_constantes)

print("\nDimensiones finales df_model_ready:", df_model_ready.shape)
```
